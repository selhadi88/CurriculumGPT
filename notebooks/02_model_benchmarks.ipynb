# Model Benchmarks
Compare BGEBiEncoder vs CurriculumTransformer on a held-out sample.

```python
import sys
sys.path.insert(0, '..')
import torch
from ml.models.transformer import CurriculumTransformer
from ml.models.bi_encoder import BGEBiEncoder

# Test CurriculumTransformer (no internet needed)
model = CurriculumTransformer()
print(f'CurriculumTransformer params: {sum(p.numel() for p in model.parameters()):,}')
```

```python
# Quick alignment score demo with CurriculumTransformer
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')

curriculum = 'Python programming, machine learning, deep learning, neural networks, PyTorch'
job_match  = 'Machine Learning Engineer: requires Python, PyTorch, deep learning, model deployment'
job_no_match = 'Marketing Manager: requires communication, brand strategy, social media, content creation'

def score(curr, job):
    ce = tokenizer(curr, return_tensors='pt', truncation=True, max_length=128, padding=True)
    je = tokenizer(job, return_tensors='pt', truncation=True, max_length=128, padding=True)
    with torch.no_grad():
        out = model(ce['input_ids'], ce['attention_mask'], je['input_ids'], je['attention_mask'])
    return out['alignment_score'].item()

print(f'Match score:    {score(curriculum, job_match):.3f}')
print(f'No-match score: {score(curriculum, job_no_match):.3f}')
```
