# Data Exploration
Explore the processed courses and jobs datasets.

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
from data.preprocessing.pipeline import load_processed

courses = load_processed('courses')
jobs = load_processed('jobs')

print(f'Courses: {len(courses):,} rows')
print(f'Jobs:    {len(jobs):,} rows')
courses.head()
```

```python
# Description length distributions
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
courses['description'].str.len().hist(bins=50, ax=axes[0], color='#6366f1')
axes[0].set_title('Course description lengths')
jobs['description'].str.len().hist(bins=50, ax=axes[1], color='#10b981')
axes[1].set_title('Job description lengths')
plt.tight_layout()
```

```python
# Category / domain breakdown
if 'category' in courses.columns:
    courses['category'].value_counts().head(15).plot(kind='barh', figsize=(10, 5), color='#6366f1')
    plt.title('Top course categories')
    plt.tight_layout()
    plt.show()

if 'domain' in jobs.columns:
    jobs['domain'].value_counts().head(15).plot(kind='barh', figsize=(10, 5), color='#10b981')
    plt.title('Top job domains')
    plt.tight_layout()
```
